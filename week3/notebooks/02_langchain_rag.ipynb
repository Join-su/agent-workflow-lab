## 전체 코드 연결
`week3/app.py`의 `policy_retrieval_chain`은 LangChain Runnable입니다. 이 Notebook은 Graph가 쓰기 전의 retrieval 결과만 작게 관찰합니다.

```python
from week3.app import POLICY_CHUNKS, policy_retrieval_chain

# POLICY_CHUNKS: splitter가 만든 정책 문서 조각 목록이다.
first_chunk = POLICY_CHUNKS[0]
(first_chunk.page_content, first_chunk.metadata)
```

```python
# policy_retrieval_chain: 질문/업무 목적을 받아 근거 Document 목록을 반환한다.
evidence = policy_retrieval_chain.invoke('정책 근거를 확인해 주세요')
# metadata는 나중에 citation으로 변환된다.
[(doc.page_content, doc.metadata['chunk_id']) for doc in evidence]
```
