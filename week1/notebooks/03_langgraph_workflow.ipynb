## 전체 코드 연결
`week1/app.py`의 `WORKFLOW`는 `StateGraph`를 compile한 객체입니다. 이 Notebook은 node를 재구현하지 않고 입력 state와 최종 state를 비교합니다.

```python
from week1.app import WORKFLOW

# initial_state: Graph가 처리하기 전의 최소 업무 상태다.
initial_state = {'question': '휴가 신청은 며칠 전에 해야 하나요?', 'agents_run': []}
final_state = WORKFLOW.invoke(initial_state)
final_state['agents_run']
```

```python
# final_state: 각 node가 더한 정보가 합쳐진 결과다.
# status/route/revision_count 중 존재하는 값만 골라 관찰한다.
{key: final_state.get(key) for key in ['status', 'route', 'revision_count', 'reason']}
```
