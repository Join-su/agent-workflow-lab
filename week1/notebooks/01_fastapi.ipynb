업무 담당자가 한 건의 요청을 API에 전달하고 결과를 확인합니다. 이 Notebook은 서비스 입구만 연습합니다.

## 전체 코드 연결
`week1/app.py`의 `create_app()`은 HTTP 경계를 만들고, endpoint는 내부 workflow를 호출합니다. Graph 내부 구현은 다음 Notebook에서 다룹니다.

```python
from fastapi.testclient import TestClient
from week1.app import create_app

# create_app(): 학습용 FastAPI 서비스를 만든다.
client = TestClient(create_app())
# payload: 현업 사용자가 보내는 최소 입력 fixture다.
payload = {'question': '휴가 신청은 며칠 전에 해야 하나요?'}
response = client.post('/query', json=payload)
response.status_code
```

```python
body = response.json()
# body: API가 내부 workflow 결과를 외부에 전달한 응답이다.
{field: body.get(field) for field in ['status', 'answer', 'citations', 'agents_run']}
```
